```python
def get_label_rotation(angle, offset):
    # Rotation must be specified in degrees :(
    rotation = np.rad2deg(angle + offset)
    if angle <= np.pi:
        alignment = "right"
        rotation = rotation + 180
    else: 
        alignment = "left"
    return rotation, alignment
```

```python
def add_labels(angles, values, labels, offset, ax):
    
    # This is the space between the end of the bar and the label
    padding = 4
    
    # Iterate over angles, values, and labels, to add all of them.
    for angle, value, label, in zip(angles, values, labels):
        angle = angle
        
        # Obtain text rotation and alignment
        rotation, alignment = get_label_rotation(angle, offset)

        # And finally add the text
        ax.text(
            x=angle, 
            y=value + padding, 
            s=label, 
            ha=alignment, 
            va="center", 
            rotation=rotation, 
            rotation_mode="anchor"
        ) 
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define font properties
plt.rcParams['font.family'] = 'Times New Roman'  # You can change to 'Times New Roman', 'Arial', etc.
plt.rcParams['font.size'] = 16              # Base font size
plt.rcParams['axes.titlesize'] = 16           # Title font size
plt.rcParams['legend.fontsize'] = 14          # Legend font size

# Define categories and ML methods
categories = ['Interpretability', 'Generalisation', 'Online Processing']
ml_methods = ['SL', 'SSL', 'UL', 'RL']

# Values for each ML method: Low=1, Moderate=2, High=3
scores = {
    'SL': [3, 1, 1],
    'SSL': [3, 2, 1],
    'UL': [2, 3, 3],
    'RL': [1, 3, 1]
}

# Compute angles
num_vars = len(categories)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]

# Radar chart setup
fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))
ax.set_theta_offset(np.pi / 2)        # Rotate so first axis is at top
ax.set_theta_direction(-1)            # Clockwise

# Plot each line
for method, values in scores.items():
    values += values[:1]
    ax.plot(angles, values, label=method, linewidth=2)
    ax.fill(angles, values, alpha=0.1)

# Custom label rotation and positioning
labels = categories
ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels, fontsize=15)

# Adjust individual label rotation for better spacing
for label, angle in zip(ax.get_xticklabels(), angles[:-1]):
    label.set_rotation(np.degrees(angle) - 90)
    label.set_horizontalalignment('center')

# Radial axis settings
# Remove default ytick labels
ax.set_yticks([1, 2, 3])
ax.set_yticklabels([])  # Hide default labels

# Custom ytick label positions
yticks = [1, 2, 3]
labels = ['Low', 'Moderate', 'High']
for y, label in zip(yticks, labels):
    ax.text(np.pi / 3, y, label, ha='left', va='center', fontsize=14)  # Adjust 'ha' and 'va' as needed

ax.set_ylim(0, 3)
ax.legend(loc='upper right', bbox_to_anchor=(1.5, 1.1), frameon=False)

plt.tight_layout()
plt.savefig("performance_comparison.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
